```python
#calculates Descriptive Statistics, Percentiles, Interquartile Range (IQR), and Outlier Detection boundaries for each numerical column in your dataset
descriptive=pd.DataFrame(index=["Mean","Median","Mode","Q1:25%","Q2:50%","Q3:75%","99%","Q4:100%","IQR","1.5rule","Lesser","Greater","min","max"],columns=quan)
for columnName in quan:
        descriptive.loc["Mean", columnName] = dataset[columnName].mean()
        descriptive.loc["Median", columnName] = dataset[columnName].median()
        descriptive.loc["Mode", columnName] = dataset[columnName].mode()[0]
        descriptive.loc["Q1:25%", columnName] = dataset.describe().loc["25%", columnName]
        descriptive.loc["Q2:50%", columnName] = dataset.describe().loc["50%", columnName]
        #descriptive.loc["Q3:75%", columnName] = dataset.describe().loc["75%", columnName]
        #descriptive.loc["1.5rule", columnName] = dataset.describe().loc["Q2:50%", columnName]
        #descriptive.loc["Q1:25%", columnName] = dataset.describe().loc["01:25%",columnName]
        #descriptive.loc["Q2:50%", columnName] = dataset.describe().loc["Q2:50%", columnName]
        descriptive.loc["Q3:75%", columnName] = dataset.describe().loc["75%", columnName]
        descriptive.loc["Q4:100%", columnName] = dataset.describe().loc["max", columnName]
        descriptive.loc["99%", columnName] = np.percentile(dataset[columnName],99)
        #descriptive.loc["MAXV", columnName] = dataset.describe().loc["max", columnName]
        #descriptive.loc["MINV", columnName] = dataset.describe().loc["min", columnName]
        descriptive.loc["IQR", columnName] =descriptive.loc["Q3:75%", columnName]-descriptive.loc["Q2:50%", columnName]
        descriptive.loc["1.5rule", columnName] =1.5*descriptive.loc["IQR",columnName]
        descriptive.loc["Lesser",columnName]=descriptive.loc["Q1:25%",columnName]-(1.5)*descriptive.loc["IQR",columnName]
        descriptive.loc["Greater",columnName]=descriptive.loc["Q3:75%",columnName]+(1.5)*descriptive.loc["IQR",columnName]
        descriptive.loc["min", columnName] = dataset.describe().loc["min", columnName]
        descriptive.loc["max", columnName] = dataset.describe().loc["max", columnName]
```

```python
# we need the new dataframe used for replace the outlier using the function
def descriptivefunc(dataset,quan):
    descriptive=pd.DataFrame(index=["Mean","Median","Mode","Q1:25%","Q2:50%","Q3:75%","99%","Q4:100%","IQR","1.5rule","Lesser","Greater","min","max"],columns=quan)
    for columnName in quan:
        descriptive.loc["Mean", columnName] = dataset[columnName].mean()
        descriptive.loc["Median", columnName] = dataset[columnName].median()
        descriptive.loc["Mode", columnName] = dataset[columnName].mode()[0]
        descriptive.loc["Q1:25%", columnName] = dataset.describe().loc["25%", columnName]
        descriptive.loc["Q2:50%", columnName] = dataset.describe().loc["50%", columnName]
        #descriptive.loc["Q3:75%", columnName] = dataset.describe().loc["75%", columnName]
        #descriptive.loc["1.5rule", columnName] = dataset.describe().loc["Q2:50%", columnName]
        #descriptive.loc["Q1:25%", columnName] = dataset.describe().loc["01:25%",columnName]
        #descriptive.loc["Q2:50%", columnName] = dataset.describe().loc["Q2:50%", columnName]
        descriptive.loc["Q3:75%", columnName] = dataset.describe().loc["75%", columnName]
        descriptive.loc["Q4:100%", columnName] = dataset.describe().loc["max", columnName]
        descriptive.loc["99%", columnName] = np.percentile(dataset[columnName],99)
        #descriptive.loc["MAXV", columnName] = dataset.describe().loc["max", columnName]
        #descriptive.loc["MINV", columnName] = dataset.describe().loc["min", columnName]
        descriptive.loc["IQR", columnName] =descriptive.loc["Q3:75%", columnName]-descriptive.loc["Q2:50%", columnName]
        descriptive.loc["1.5rule", columnName] =1.5*descriptive.loc["IQR",columnName]
        descriptive.loc["Lesser",columnName]=descriptive.loc["Q1:25%",columnName]-(1.5)*descriptive.loc["IQR",columnName]
        descriptive.loc["Greater",columnName]=descriptive.loc["Q3:75%",columnName]+(1.5)*descriptive.loc["IQR",columnName]
        descriptive.loc["min", columnName] = dataset.describe().loc["min", columnName]
        descriptive.loc["max", columnName] = dataset.describe().loc["max", columnName]
    return descriptive
```

```python
# Call the newly created function and pass the parameters to check for outliers and replace them with the minimum value if found.
descriptivenew=descriptivefunc(dataset,quan)
```
